```python
# ============================================================
# EXOPLANET HABITABILITY PREDICTION
# DATA LOADING + VALIDATION + FEATURE SELECTION + CLEANING
# ============================================================

import pandas as pd
from sklearn.impute import SimpleImputer

# ------------------------------------------------------------
# 1. Load NASA Exoplanet Archive Dataset
# ------------------------------------------------------------

file_path = r"D:\Exoplanet Habitability Prediction\dataset\exoplanets.xlsx"

df = pd.read_excel(
    file_path,
    header=292
)

print("=" * 60)
print("DATASET LOADED SUCCESSFULLY")
print("=" * 60)

print("Original shape:", df.shape)


# ------------------------------------------------------------
# 2. Basic Dataset Information
# ------------------------------------------------------------

print("\nNumber of rows:", df.shape[0])
print("Number of columns:", df.shape[1])

print("\nFirst 5 rows:")
display(df.head())


# ------------------------------------------------------------
# 3. Duplicate Check
# ------------------------------------------------------------

duplicate_count = df.duplicated().sum()

print("\nDuplicate rows:", duplicate_count)


# ------------------------------------------------------------
# 4. Missing Value Analysis
# ------------------------------------------------------------

missing_info = pd.DataFrame({
    "Missing_Count": df.isnull().sum(),
    "Missing_Percentage": (
        df.isnull().sum() / len(df) * 100
    )
})

print("\nTop 20 columns with missing values:")
display(
    missing_info
    .sort_values("Missing_Percentage", ascending=False)
    .head(20)
)


# ------------------------------------------------------------
# 5. Select Relevant Exoplanet Features
# ------------------------------------------------------------

candidate_features = [
    "pl_orbper",     # Orbital period
    "pl_orbsmax",    # Orbital semi-major axis
    "pl_rade",       # Planet radius
    "pl_masse",      # Planet mass
    "pl_dens",       # Planet density
    "pl_insol",      # Insolation
    "pl_eqt",        # Equilibrium temperature
    "st_teff",       # Stellar effective temperature
    "st_rad",        # Stellar radius
    "st_mass",       # Stellar mass
    "st_met",        # Stellar metallicity
    "st_logg"        # Stellar surface gravity
]

print("\nCandidate features:")
print(candidate_features)


# ------------------------------------------------------------
# 6. Check Whether Features Exist
# ------------------------------------------------------------

available_features = [
    column
    for column in candidate_features
    if column in df.columns
]

print("\nAvailable candidate features:")
print(available_features)


# ------------------------------------------------------------
# 7. Create ML Working Dataset
# ------------------------------------------------------------

model_df = df[available_features].copy()


# ------------------------------------------------------------
# 8. Convert Features to Numeric
# ------------------------------------------------------------

for column in available_features:
    model_df[column] = pd.to_numeric(
        model_df[column],
        errors="coerce"
    )


# ------------------------------------------------------------
# 9. Check Missing Values in Selected Features
# ------------------------------------------------------------

feature_missing = pd.DataFrame({
    "Missing_Count": model_df.isnull().sum(),
    "Missing_Percentage":
        model_df.isnull().sum() / len(model_df) * 100
})

print("\nMissing values in selected features:")
display(feature_missing)


# ------------------------------------------------------------
# 10. Keep Features with at Least 70% Available Data
# ------------------------------------------------------------

usable_features = [
    column
    for column in available_features
    if model_df[column].notna().mean() >= 0.70
]

print("\nUsable features:")
print(usable_features)


# ------------------------------------------------------------
# 11. Keep Only Usable Features
# ------------------------------------------------------------

model_df = model_df[usable_features].copy()


# ------------------------------------------------------------
# 12. Median Imputation
# ------------------------------------------------------------

imputer = SimpleImputer(strategy="median")

model_df[usable_features] = imputer.fit_transform(
    model_df[usable_features]
)


# ------------------------------------------------------------
# 13. Remove Duplicate Records
# ------------------------------------------------------------

before = len(model_df)

model_df = model_df.drop_duplicates()

after = len(model_df)

print("\nDuplicates removed:", before - after)


# ------------------------------------------------------------
# 14. Final Dataset Check
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("CLEANING COMPLETED")
print("=" * 60)

print("Final shape:", model_df.shape)

print("\nFinal features:")
print(model_df.columns.tolist())

print("\nRemaining missing values:")
print(model_df.isnull().sum())

print("\nFinal dataset preview:")
display(model_df.head())
```

```python
# STEP 3: Basic Dataset Cleaning

import pandas as pd
import numpy as np

# Check dataset shape
print("Dataset shape:", df.shape)

# Check data types
print("\nData types:")
print(df.dtypes)

# Count missing values
missing = df.isnull().sum()

# Show only columns having missing values
print("\nColumns with missing values:")
print(missing[missing > 0].sort_values(ascending=False).head(30))

# Check duplicate rows
print("\nDuplicate rows:", df.duplicated().sum())

# Remove duplicate rows
df = df.drop_duplicates()

print("\nShape after removing duplicates:", df.shape)
```

```python
# STEP 4: Select Important Features

features = [
    "pl_name",
    "pl_rade",        # Planet radius
    "pl_bmasse",      # Planet mass
    "pl_orbsmax",     # Orbital distance
    "pl_orbper",      # Orbital period
    "pl_eqt",         # Equilibrium temperature
    "pl_insol",       # Stellar insolation
    "st_teff",        # Star temperature
    "st_rad",         # Star radius
    "st_mass",        # Star mass
    "st_lum",         # Star luminosity
    "st_logg",        # Surface gravity
    "sy_dist"         # Distance from Earth
]

# Keep only columns that actually exist in the dataset
available_features = [col for col in features if col in df.columns]

df_model = df[available_features].copy()

print("Selected features:")
print(available_features)

print("\nNew dataset shape:", df_model.shape)

print("\nFirst 5 rows:")
display(df_model.head())
```

```python
# STEP 5: Missing Value Analysis

print("Missing values in selected features:\n")

missing_values = df_model.isnull().sum()

missing_percentage = (missing_values / len(df_model)) * 100

missing_summary = pd.DataFrame({
    "Missing Values": missing_values,
    "Missing Percentage": missing_percentage.round(2)
})

print(missing_summary.sort_values("Missing Percentage", ascending=False))

print("\nTotal rows:", len(df_model))
```

```python
# STEP 6: Select features with better data availability

model_features = [
    "pl_rade",
    "pl_orbper",
    "st_teff",
    "st_rad",
    "st_mass",
    "st_logg",
    "sy_dist"
]

# Create a new dataframe
model_df = df[model_features].copy()

# Remove rows where important values are missing
model_df = model_df.dropna()

print("Original rows:", len(df))
print("Rows after cleaning:", len(model_df))
print("Columns:", model_df.columns.tolist())

print("\nMissing values after cleaning:")
print(model_df.isnull().sum())

print("\nFirst 5 rows:")
display(model_df.head())
```

```python
# STEP 7: Prepare features for habitability prediction

model_features = [
    "pl_rade",
    "pl_orbper",
    "pl_eqt",
    "pl_insol",
    "st_teff",
    "st_rad",
    "st_mass",
    "st_logg"
]

model_df = df[model_features].copy()

# Remove rows missing the values needed for our model
model_df = model_df.dropna()

print("Original dataset:", df.shape)
print("Clean dataset:", model_df.shape)

print("\nMissing values:")
print(model_df.isnull().sum())

display(model_df.head())
```

```python
# STEP 8: Create Habitability Target

model_df["habitable"] = (
    (model_df["pl_rade"] >= 0.5) &
    (model_df["pl_rade"] <= 2.0) &
    (model_df["pl_eqt"] >= 200) &
    (model_df["pl_eqt"] <= 320) &
    (model_df["pl_insol"] >= 0.25) &
    (model_df["pl_insol"] <= 2.0)
).astype(int)

print("Habitability labels created successfully.")

print("\nLabel distribution:")
print(model_df["habitable"].value_counts())

print("\nPercentage distribution:")
print(model_df["habitable"].value_counts(normalize=True).mul(100).round(2))

display(model_df.head())
```

```python
# STEP 9: Prepare Data for Machine Learning

from sklearn.model_selection import train_test_split

# Features (input)
X = model_df.drop("habitable", axis=1)

# Target (output)
y = model_df["habitable"]

# Split dataset into training and testing data
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Total samples:", len(X))
print("Training samples:", len(X_train))
print("Testing samples:", len(X_test))

print("\nTraining target distribution:")
print(y_train.value_counts())

print("\nTesting target distribution:")
print(y_test.value_counts())
```

```python
# STEP 10: Train Random Forest Model

from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

# Create Random Forest model
rf_model = RandomForestClassifier(
    n_estimators=100,
    random_state=42,
    class_weight="balanced"
)

# Train the model
rf_model.fit(X_train, y_train)

# Make predictions
y_pred = rf_model.predict(X_test)

# Accuracy
accuracy = accuracy_score(y_test, y_pred)

print("Random Forest Model")
print("-------------------")
print("Accuracy:", round(accuracy * 100, 2), "%")

print("\nClassification Report:")
print(classification_report(y_test, y_pred))

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred))
```

```python
# STEP 11: Feature Importance

import pandas as pd
import matplotlib.pyplot as plt

feature_importance = pd.DataFrame({
    "Feature": X_train.columns,
    "Importance": rf_model.feature_importances_
})

feature_importance = feature_importance.sort_values(
    by="Importance",
    ascending=False
)

print("Feature Importance:")
display(feature_importance)

# Plot feature importance
plt.figure(figsize=(10, 6))

plt.barh(
    feature_importance["Feature"],
    feature_importance["Importance"]
)

plt.xlabel("Importance")
plt.ylabel("Feature")
plt.title("Random Forest Feature Importance")
plt.gca().invert_yaxis()

plt.show()
```

```python
# STEP 12: Confusion Matrix Visualization

import matplotlib.pyplot as plt
from sklearn.metrics import ConfusionMatrixDisplay

# Create confusion matrix display
ConfusionMatrixDisplay.from_predictions(
    y_test,
    y_pred,
    display_labels=["Not Habitable", "Potentially Habitable"]
)

plt.title("Random Forest Confusion Matrix")
plt.xlabel("Predicted Label")
plt.ylabel("Actual Label")
plt.show()
```

```python
# STEP 13: Train Logistic Regression Model

from sklearn.linear_model import LogisticRegression
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

# Create pipeline
lr_model = Pipeline([
    ("scaler", StandardScaler()),
    ("classifier", LogisticRegression(
        max_iter=1000,
        class_weight="balanced",
        random_state=42
    ))
])

# Train model
lr_model.fit(X_train, y_train)

# Make predictions
lr_pred = lr_model.predict(X_test)

# Calculate accuracy
lr_accuracy = accuracy_score(y_test, lr_pred)

print("Logistic Regression Model")
print("------------------------")
print("Accuracy:", round(lr_accuracy * 100, 2), "%")

print("\nClassification Report:")
print(classification_report(y_test, lr_pred))

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, lr_pred))
```

```python
# STEP 14: Compare ML Models

from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score
import pandas as pd
import matplotlib.pyplot as plt

# Random Forest metrics
rf_accuracy = accuracy_score(y_test, y_pred)
rf_precision = precision_score(y_test, y_pred, zero_division=0)
rf_recall = recall_score(y_test, y_pred, zero_division=0)
rf_f1 = f1_score(y_test, y_pred, zero_division=0)

# Logistic Regression metrics
lr_accuracy = accuracy_score(y_test, lr_pred)
lr_precision = precision_score(y_test, lr_pred, zero_division=0)
lr_recall = recall_score(y_test, lr_pred, zero_division=0)
lr_f1 = f1_score(y_test, lr_pred, zero_division=0)

# Create comparison table
comparison = pd.DataFrame({
    "Model": ["Random Forest", "Logistic Regression"],
    "Accuracy": [rf_accuracy, lr_accuracy],
    "Precision": [rf_precision, lr_precision],
    "Recall": [rf_recall, lr_recall],
    "F1 Score": [rf_f1, lr_f1]
})

print("Model Comparison:")
display(comparison.round(4))

# Accuracy graph
plt.figure(figsize=(8, 5))

plt.bar(
    comparison["Model"],
    comparison["Accuracy"] * 100
)

plt.ylabel("Accuracy (%)")
plt.xlabel("Model")
plt.title("Machine Learning Model Comparison")
plt.ylim(0, 105)

plt.show()
```

```python
# STEP 15: Exoplanet Habitability Prediction System

def predict_habitability():
    
    print("=== Exoplanet Habitability Prediction ===")
    
    # Take input from user
    pl_rade = float(input("Planet Radius (Earth radii): "))
    pl_orbper = float(input("Orbital Period (days): "))
    pl_eqt = float(input("Equilibrium Temperature (K): "))
    pl_insol = float(input("Stellar Insolation (Earth units): "))
    st_teff = float(input("Star Temperature (K): "))
    st_rad = float(input("Star Radius (Solar radii): "))
    st_mass = float(input("Star Mass (Solar masses): "))
    st_logg = float(input("Star Surface Gravity: "))
    
    # Create input dataframe
    new_planet = pd.DataFrame([{
        "pl_rade": pl_rade,
        "pl_orbper": pl_orbper,
        "pl_eqt": pl_eqt,
        "pl_insol": pl_insol,
        "st_teff": st_teff,
        "st_rad": st_rad,
        "st_mass": st_mass,
        "st_logg": st_logg
    }])
    
    # Prediction
    prediction = rf_model.predict(new_planet)[0]
    
    # Probability
    probability = rf_model.predict_proba(new_planet)[0]
    
    print("\nPrediction:")
    
    if prediction == 1:
        print("Potentially Habitable")
    else:
        print("Not Potentially Habitable")
    
    print("\nPrediction probabilities:")
    print("Not Habitable:", round(probability[0] * 100, 2), "%")
    print("Potentially Habitable:", round(probability[1] * 100, 2), "%")


# Run prediction system
predict_habitability()
```
